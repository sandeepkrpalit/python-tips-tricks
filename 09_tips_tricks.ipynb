### 9. Index of the Biggest Number

Using the enumerator(), max() and min() functions, we can find the index of the biggest and the smallest number in a list. The max function is high-order function; it takes another function as an argument.

Find the index of the largest number

```python
x = [12, 45, 67, 89, 34, 67, 13]

max_num = max(enumerate(x, start = 0),)
```
